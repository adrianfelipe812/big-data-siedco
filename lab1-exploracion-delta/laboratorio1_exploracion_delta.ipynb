##5. Almacenamiento en formato tabular
Convierte o registra el dataset como una tabla dentro del entorno de Databricks, preferiblemente en formato Delta. Explica por qué este tipo de almacenamiento resulta útil en escenarios de Big Data y qué ventajas ofrece frente a trabajar solo con archivos sueltos.


###5.1 Crear la tabla Delta

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.big_data.coin_data_delta")
```

###5.2 Comprobación de la tabla

```python
spark.sql("""SHOW TABLES IN workspace.big_data""").show()
```

### 5.3 Consulta de la nueva tabla

```python
df_delta = spark.table("workspace.big_data.coin_data_delta")

display(df_delta.limit(10))
```

###5.4 Verificar que conserva los registros

```python
print(f"Registros originales: {df.count()}")
print(f"Registros en Delta: {df_delta.count()}")
```

###5.5 Comprobacion Delta

```python
spark.sql("""DESCRIBE DETAIL workspace.big_data.coin_data_delta""").show(truncate=False)
```

###5.6 Explicación del almacenamiento Delta
Después de realizar la exploración inicial, el conjunto de datos fue almacenado en una nueva tabla utilizando el formato Delta. La tabla generada se denomina `workspace.big_data.coin_data_delta`.

El formato Delta resulta especialmente útil en escenarios de Big Data porque permite gestionar grandes volúmenes de información de una manera más confiable y eficiente que el almacenamiento basado únicamente en archivos CSV.

Entre sus principales ventajas se encuentran el soporte para transacciones ACID, una mayor consistencia de los datos, capacidades de actualización y eliminación de registros, conservación del historial de cambios y una integración nativa con Apache Spark y Databricks.

A diferencia de trabajar únicamente con archivos CSV, una tabla Delta proporciona una estructura de almacenamiento más robusta para procesos de análisis y transformación de datos. Además, facilita la construcción de procesos de ingeniería de datos en los que se requiere modificar, actualizar o consultar grandes cantidades de información.

En este ejercicio, el archivo CSV utilizado inicialmente quedó registrado en Databricks como workspace.big_data.coin_data y posteriormente fue almacenado en formato Delta como workspace.big_data.coin_data_delta, permitiendo disponer de los mismos datos en un formato más apropiado para posteriores procesos de Big Data.

##Paso 6. Consulta básica de datos
Ejecuta consultas básicas sobre la tabla creada. Como mínimo, debes realizar:

1. una consulta para contar el número total de registros
2. una consulta para seleccionar columnas relevantes
3. una consulta con filtro
4. una consulta agregada simple, por ejemplo promedio, suma o conteo por categoría

Puedes desarrollar estas consultas en SQL o en PySpark, según la orientación del docente

###6.1 Contar el número total de registros

```python
%sql
SELECT COUNT(*) AS total_registros
FROM workspace.big_data.coin_data_delta;
```

###6.2 Seleccionar columnas relevantes

```python
%sql
SELECT 
    coin,
    date,
    price,
    total_volume
FROM workspace.big_data.coin_data_delta
LIMIT 10;
```

###6.3 Consulta con filtro
Ejemplo consulta de registros de Bitcoin y filtro de precios

```python
%sql
SELECT 
    coin,
    date,
    price,
    market_cap,
    total_volume
FROM workspace.big_data.coin_data_delta
WHERE coin = 'bitcoin'
ORDER BY date;
```

```python
%sql
SELECT 
    coin,
    date,
    price
FROM workspace.big_data.coin_data_delta
WHERE price > 120000
ORDER BY price DESC;
```

###6.4 Consulta agregada
Precio promedio por criptomoneda

```python
%sql
SELECT 
    coin,
    AVG(price) AS precio_promedio
FROM workspace.big_data.coin_data_delta
GROUP BY coin
ORDER BY precio_promedio DESC;
```

Una vez creada la tabla en formato Delta, se realizaron diferentes consultas básicas utilizando **SQL** directamente sobre la tabla `workspace.big_data.coin_data_delta`.

En primer lugar, se realizó un conteo del número total de registros para verificar el volumen de información almacenado. Posteriormente, se seleccionaron columnas relevantes del dataset para consultar información específica de las criptomonedas.

También se realizó una consulta utilizando un filtro sobre la columna `coin`, permitiendo obtener únicamente los registros correspondientes y filtros de precios a Bitcoin. Finalmente, se ejecutó una consulta agregada para calcular el precio promedio de cada criptomoneda mediante la función `AVG()` y la agrupación por categoría utilizando `GROUP BY`.

Estas consultas permiten comprobar el funcionamiento de la tabla Delta y realizar operaciones básicas de análisis sobre los datos almacenados en Databricks.

## 8. Análisis de la experiencia de uso

#### ¿Qué ventajas observamos al trabajar en una plataforma unificada como Databricks?

Durante la práctica pudimos observar de manera más clara cómo Databricks permite integrar diferentes etapas de un proceso de análisis de datos en un mismo entorno. Partimos desde la carga del archivo CSV y posteriormente realizamos su exploración mediante Apache Spark, almacenamiento en formato Delta, consultas SQL y finalmente una visualización de los datos, sin tener que trasladar la información entre diferentes plataformas.

Una de las principales ventajas que encontramos fue precisamente esta integración. Tener en un mismo espacio las herramientas para cargar, procesar, almacenar, consultar y visualizar los datos hace que el trabajo sea más organizado y facilita el seguimiento de cada etapa. También nos pareció útil poder trabajar tanto con PySpark como con SQL, ya que cada uno puede resultar más conveniente dependiendo de la tarea que se esté realizando.

#### ¿Qué dificultades encontramos en la carga, organización o consulta de los datos?

Durante la práctica encontramos algunas dificultades relacionadas principalmente con la familiarización con la plataforma y la forma en que Databricks organiza los datos. Por ejemplo, tuvimos que diferenciar entre el archivo CSV original y la tabla registrada en el catálogo, además de comprender la relación entre el esquema `big_data`, la tabla `workspace.big_data.coin_data` y posteriormente la tabla Delta `workspace.big_data.coin_data_delta`.

También encontramos que es importante prestar atención al lenguaje seleccionado para cada celda. Una consulta escrita en SQL, por ejemplo, genera un error si se ejecuta dentro de una celda configurada para Python. Aunque fueron dificultades sencillas de solucionar, nos permitieron entender que trabajar con una plataforma de datos no consiste únicamente en conocer los comandos, sino también en comprender cómo está organizado el entorno y qué herramienta utilizar en cada etapa.

#### ¿Qué papel cumplen el almacenamiento estructurado y el procesamiento distribuido en un contexto de datos masivos?

Consideramos que el almacenamiento estructurado cumple un papel fundamental porque permite organizar la información de una manera más adecuada para su consulta y procesamiento. En nuestro caso, partir de un archivo CSV fue suficiente para cargar los datos, pero convertirlos posteriormente en una tabla Delta permitió disponer de una estructura más robusta para realizar consultas y futuras transformaciones.

Por otro lado, el procesamiento distribuido cobra mayor importancia a medida que aumenta el volumen de información. Aunque el dataset utilizado durante esta práctica no representa por sí solo un volumen masivo, nos permitió comprender el concepto detrás de tecnologías como Apache Spark: distribuir las operaciones de procesamiento entre diferentes recursos para poder trabajar de manera eficiente con conjuntos de datos mucho más grandes.

Esto nos permitió entender que herramientas como Spark no solo son útiles cuando ya se tienen grandes cantidades de datos, sino que también sirven para familiarizarnos con una forma de procesamiento que puede escalar a medida que las necesidades de una organización aumentan.

Consideramos que la experiencia tiene una relación directa con situaciones que pueden presentarse en escenarios reales de analítica de datos. Entre estas etapas se encuentran la carga, organización, transformación, consulta y visualización de los datos.

En este sentido, Databricks permite integrar gran parte de este flujo dentro de una misma plataforma, lo que puede ser especialmente útil en proyectos de ingeniería de datos, Business Intelligence y analítica avanzada.
